# UC Admissions Data Challenge Question Sprint

```python
from pathlib import Path
import pandas as pd

DATA = Path('given_materials/data')
if not DATA.exists():
    DATA = Path('../given_materials/data')
```

```python
# 1
df = pd.read_csv(DATA / 'bay_area_modeling_table.csv')
df = df[df.fall_term == 2025]
answer = df[df.campus != 'Universitywide'].applicants.sum() / df[df.campus == 'Universitywide'].applicants.sum()
print(round(answer, 2))
```

```python
# 2
df = pd.read_csv(DATA / 'bay_area_modeling_table.csv')
df = df[(df.fall_term == 2025) & (df.campus == 'Los Angeles')]
answer = df.admits.sum() / df.applicants.sum()
print(round(answer, 4))
```

```python
# 3
df = pd.read_csv(DATA / 'uc_freshman_admission_by_discipline.csv')
df = df[df.fall_term == 2025]
overall = df[df.broad_discipline == 'All disciplines'][['campus', 'admit_rate']]
cs = df[df.broad_discipline == 'Computer Science'][['campus', 'admit_rate']]
merged = overall.merge(cs, on='campus', suffixes=('_overall', '_cs'))
answer = merged.assign(cost=merged.admit_rate_overall - merged.admit_rate_cs).sort_values('cost').iloc[-1].campus
print(answer)
```

```python
# 4
df = pd.read_csv(DATA / 'uc_freshman_admission_by_discipline.csv')
row = df[(df.fall_term == 2025) & (df.campus == 'Berkeley') & (df.broad_discipline == 'Computer Science')].iloc[0]
answer = row.admit_gpa_p75 - row.admit_gpa_p25
print(round(answer, 2))
```

```python
# 5
df = pd.read_csv(DATA / 'uc_admissions_summary_by_ethnicity.csv')
df = df[(df.fall_term == 2025) & (df.entrant_level == 'freshman') & df.ethnicity.isin(['White', 'Hispanic/Latino(a)'])]
rates = df.pivot_table(index=['campus', 'ethnicity'], columns='count_type', values='n').reset_index()
rates['admit_rate'] = rates.Adm / rates.App
rates = rates[rates.campus != 'Systemwide'].pivot(index='campus', columns='ethnicity', values='admit_rate')
answer = (rates.White > rates['Hispanic/Latino(a)']).sum()
print(int(answer))
```

```python
# 6
df = pd.read_csv(DATA / 'uc_admissions_summary_by_ethnicity.csv')
df = df[(df.fall_term == 2025) & (df.entrant_level == 'freshman') & (df.campus == 'Systemwide') & df.ethnicity.isin(['White', 'Hispanic/Latino(a)'])]
rates = df.pivot_table(index='ethnicity', columns='count_type', values='n')
rates['admit_rate'] = rates.Adm / rates.App
answer = rates.admit_rate.idxmax()
print(answer)
```

```python
# 7
df = pd.read_csv(DATA / 'bay_area_modeling_table.csv')
df = df[(df.fall_term == 2023) & (df.campus == 'Universitywide')]
answer = df.enrolled_ccc.sum() / df.hs_completers.sum()
print(round(answer, 4))
```

```python
# 8
df = pd.read_csv(DATA / 'bay_area_modeling_table.csv')
row = df[(df.fall_term == 2023) & (df.campus == 'Universitywide') & (df.high_school == 'MISSION SAN JOSE HIGH SCHOOL')].iloc[0]
answer = row.applicants / row.ag_completers
print(round(answer, 4))
```

```python
# 9
df = pd.read_csv(DATA / 'bay_area_modeling_table.csv')
df = df[(df.fall_term == 2025) & (df.campus == 'Universitywide') & (df.applicants >= 1)]
ids = df.cds_code.fillna(df.high_school + '|' + df.city + '|' + df.county)
print(int(ids.nunique()))
```

```python
# 10
schools = ['HERCULES HIGH SCHOOL', 'MISSION SENIOR HIGH SCHOOL', 'MONTEREY TRAIL HIGH SCHOOL', 'PHILLIP & SALA BURTON ACAD HS', 'RANCHO SAN JUAN HIGH SCHOOL']
df = pd.read_csv(DATA / 'dashboard_data.csv')
df = df[(df.campus == 'Berkeley') & (df.fall_term.between(2022, 2025)) & df.high_school.isin(schools)]
df = df.dropna(subset=['admit_rate_residual', 'applicants'])
df['weighted_residual'] = df.admit_rate_residual * df.applicants
scores = df.groupby('high_school')[['weighted_residual', 'applicants']].sum()
answer = (scores.weighted_residual / scores.applicants).idxmax()
print(answer)
```
